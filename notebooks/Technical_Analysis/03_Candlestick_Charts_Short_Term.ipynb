# Short-Term Candlestick Analysis

In this notebook, we visualize short-term price action of a selected stock (**TICKER**) using **candlestick charts**.

We focus on:
- **Last 1 month**
- **Last 5 trading days**

Candlestick charts provide insight into intraday price behavior, volatility, and short-term trend direction.

## Functions: Candlestick Chart Visualization

We define reusable functions to:
- Download short-term OHLC data
- Plot candlestick charts with volume
- Save figures automatically to the `figures/` directory


```python
def download_price_data(ticker, period="1mo"):
    """
    Downloads and prepares OHLCV price data for candlestick plotting.

    Parameters
    ----------
    ticker : str
        Stock symbol.
    period : str, optional
        Time period to download (e.g., '1mo', '5d').

    Returns
    -------
    pd.DataFrame
        Cleaned OHLCV price DataFrame.
    """
    df = yf.download(ticker, period=period)
    df = df.astype(float)

    # Remove timezone
    df.index = df.index.tz_localize(None)

    # Handle MultiIndex structure if present
    if isinstance(df.columns, pd.MultiIndex):
        df = df.xs(ticker, level="Ticker", axis=1)
        df.index = df.index.tz_localize(None)

    return df

```

```python
def plot_candlestick(df, ticker, title_suffix, save_path=None):
    """
    Plots a candlestick chart with volume in the notebook
    and optionally saves it to a file.

    Parameters
    ----------
    df : pd.DataFrame
        OHLCV price data.
    ticker : str
        Stock symbol.
    title_suffix : str
        Title suffix (e.g., '1 Month', '1 Week').
    save_path : str, optional
        Path to save the figure. If None, figure is not saved.
    """
    # --- Create figure object to display AND save ---
    fig, axlist = mpf.plot(
        df,
        type="candle",
        style="yahoo",
        volume=True,
        title=f"{ticker} – {title_suffix}",
        figscale=1.5,
        returnfig=True  # <-- important: returns the figure for further use
    )

    # --- Save figure if requested ---
    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.savefig(save_path, dpi=150, bbox_inches="tight")

    # --- Show figure inline in notebook ---
    fig.show()

```

## Example Usage

We now visualize short-term candlestick charts for the selected stock.


```python
# --- Load data ---
df_1mo = download_price_data(TICKER, period="1mo")
df_5d = download_price_data(TICKER, period="5d")

# --- Plot and save figures ---
plot_candlestick(
    df_1mo,
    ticker=TICKER,
    title_suffix="1 Month",
    save_path=f"figures/{TICKER}_candlestick_1_month.png"
)

plot_candlestick(
    df_5d,
    ticker=TICKER,
    title_suffix="1 Week",
    save_path=f"figures/{TICKER}_candlestick_1_week.png"
)

```
